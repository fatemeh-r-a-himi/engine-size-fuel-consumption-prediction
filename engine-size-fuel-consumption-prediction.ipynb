# 🚗 Engine Size & Fuel Consumption

A beginner-friendly **Machine Learning** project that explores the relationship between engine size and fuel consumption using **Linear Regression**.


## 1. Import Libraries


```python
# NumPy: numerical operations
import numpy as np

# Pandas: working with tabular data
import pandas as pd

# Seaborn and Matplotlib: data visualization
import seaborn as sns
import matplotlib.pyplot as plt

# Scikit-learn: train/test split and Linear Regression model
from sklearn import linear_model
from sklearn.model_selection import train_test_split
```

## 2. Load the Dataset


```python
# Load the CSV dataset into a Pandas DataFrame
# Make sure the CSV file is in the same folder as this notebook.
df = pd.read_csv('engine_size_fuel_consumption.csv')

# Display the first few rows of the dataset
df.head()
```

## 3. Explore the Data


```python
# Show basic statistical information about the numerical columns
df.describe()
```

```python
# Plot the distribution of fuel consumption values
sns.histplot(data=df, x='Fuel_Consumption_L_per_100km', kde=True)
plt.title('Fuel Consumption Distribution')
plt.xlabel('Fuel Consumption (L/100km)')
plt.ylabel('Count')
plt.show()
```

```python
# Create a correlation heatmap to see the relationship between the variables
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## 4. Select Features and Target


```python
# Feature: the input variable used by the model
feature = df.drop('Fuel_Consumption_L_per_100km', axis=1)

# Target: the value that the model will try to predict
target = df['Fuel_Consumption_L_per_100km']

# Display the feature and target data
feature.head()
target.head()
```

## 5. Split the Data into Training and Testing Sets


```python
# Split the dataset into training and testing sets
# 80% of the data is used for training and 20% for testing.
# random_state makes the split reproducible.
feature_train, feature_test, target_train, target_test = train_test_split(
    feature,
    target,
    test_size=0.2,
    random_state=42
)
```

## 6. Create the Linear Regression Model


```python
# Create a Linear Regression model
reg_simple = linear_model.LinearRegression()
```

## 7. Train the Model


```python
# Train the model using the training data
reg_simple.fit(feature_train, target_train)
```

## 8. Make Predictions


```python
# Use the trained model to predict fuel consumption for the test data
pred_test = reg_simple.predict(feature_test)

# Display the predicted values
pred_test
```

## 9. Compare Actual and Predicted Values


```python
# Plot the actual values and the predicted values
plt.figure(figsize=(8, 6))
plt.scatter(feature_test, target_test, label='Actual Values')
plt.scatter(feature_test, pred_test, label='Predicted Values')

plt.xlabel('Engine Size (L)')
plt.ylabel('Fuel Consumption (L/100km)')
plt.title('Actual vs Predicted Fuel Consumption')
plt.legend()
plt.show()
```
